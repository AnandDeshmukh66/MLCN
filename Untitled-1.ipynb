## 1. An overview of the Dataset including the Dataset Characteristics and Exploratory Data Analysis, Data Preprocessing, and performance of different models.

- **Dataset Characteristics:** CIC-IDS2017 dataset contains network traffic data for the development and evaluation of intrusion detection systems. The dataset is designed to be representative of modern network traffic and includes more than 2.8 million network packets captured over a period of seven days in a real network environment. The dataset includes normal traffic and seven different attack scenarios: Brute Force, Heartbleed, Botnet, DoS, DDoS, Web Attack and Infiltration. The dataset is highly imbalanced. The majority of records belongs to the 'Benign' class and relatively few records belongs to the other classes. The dataset consists of 2830743 rows and 79 columns. In these columns, 78 of them are features that are numerical and the 'label' column is categorical.

- **Exploratory Data Analysis:** This dataset have many duplicate values (308381), which creates bias that is not good for the machine learning model. The number of missing values and infinite values on two columns 'Flow Bytes/s' and 'Flow Packets/s' are very few comparing to the size of the dataset which is only 0.06% (1564). However, the values were handled using appropriate methods.

- **Data Preprocessing:** We dropped the duplicates, replaced any infinite values (positive or negative) with NaN (not a number) and filled the missing values with median values. Since we have a very large dataset, the initial memory usage is quite high leading to session crashes. We later worked our way through this by down casting data types based on the min and max values available. We try to reduce the memory usages that is helpful for our model.

- **Data Analysis:** We grouped similar attacks together to analyze the dataset and identify patterns in the different types of attacks. We took a sample from the population (20%). Later we did some data analysis which consist of plotting various kinds of charts, correlation matrices etc. to see the relationships between features, types of attacks present in the dataset etc. In our analysis, we noticed there are a good number of features that are strongly, even directly correlated with other features (both positive and negative). This is an issue it introduces multi-collinearity which can highly impact the machine learning models that we will develop later.

  Also, as the dataset is quite huge and has more than 70 features, it would be extremely difficult to train models using limited resources. In order to overcome this issue, we used PCA (Principal Component Analysis) to reduece dimensions. We played with the number of components to see how much information we can preserve. We monitored the '*explained_variance_ratio_*' to make sure we retain most informations. However, it was a bit challenging to reduce dimensions while preserving the information to train the models. We performed StandardScaler before performing Incremental PCA.

- **ML Models:** It is worth mentioning the fact that the following dataset is highly imbalanced. So, we created a balanced dataset out of this with our domain knowledge to train various ML models. Since our dataset is quite large and has a reasonable amount of samples to train and test different ML models using various classification algorithms (Logistic Regression, Support Vector Machine for Binary Classification and Random Forest Classifier, Decision Tree, K Nearest Neighbours for Multi-class Classification).

  For binary classifications, we trained the models to distinguish between normal traffic and anomalous traffic. This means it will only predict whether an intrusion is taking place or not. Alternatively, using the multi-class classification algorithms, we further extended our prediction capabilities to identify which type of attack or intrusion is taking place. We tried both binary classifications and multi-class classifications to see how the data holds up. Later we cross-validated, evaluated and compared those models to see which one works better or worse.

- **Performance Evaluation:** After training multiple machine learning models, we proceeded to evaluate their respective performances. Our evaluation process involved comparing the accuracy, recall, f1-score and confusion matrix of each model. Through analysis of the results, we were able to see which model performed the best and which performed worse.

```python
# ================================================================
# MLCN — 24-FEATURE ATTACK TRAFFIC CHARACTERIZATION
# ================================================================
#
# PURPOSE
# -------
# Characterize the empirical feature-space behavior of:
#   BENIGN
#   Brute Force
#   DDoS
#   DoS
#   Port Scan
#
# This cell DOES NOT:
#   - clean the dataset
#   - rebalance the dataset
#   - split the dataset
#   - train XGBoost
#   - generate network traffic
#
# It analyzes the already-prepared `data_5class` DataFrame.
#
# OUTPUTS
# -------
# 1. Per-class feature statistics
# 2. Percentile distributions
# 3. Zero/missing/infinite/skew information
# 4. BENIGN-vs-attack comparisons
# 5. Effect sizes
# 6. One-vs-BENIGN ROC-AUC
# 7. Feature correlations
# 8. Mutual-information feature importance
# 9. Attack fingerprints
# 10. Feature -> traffic-characteristic mapping
# 11. Machine-readable CSV/JSON output files
#
# ================================================================

import os
import json
import warnings
import numpy as np
import pandas as pd

from sklearn.feature_selection import mutual_info_classif
from sklearn.metrics import roc_auc_score

warnings.filterwarnings("ignore")


# ================================================================
# 1. CONFIGURATION
# ================================================================

DATAFRAME_NAME = "data_5class"
LABEL_COLUMN = "Attack Type"

FEATURES = [
    "Flow Duration",
    "Total Fwd Packets",
    "Total Backward Packets",
    "Fwd Packets Length Total",
    "Bwd Packets Length Total",
    "Fwd Packet Length Mean",
    "Bwd Packet Length Mean",
    "Flow Bytes/s",
    "Flow Packets/s",
    "Flow IAT Mean",
    "Flow IAT Std",
    "Fwd IAT Mean",
    "Bwd IAT Mean",
    "Packet Length Mean",
    "Packet Length Std",
    "FIN Flag Count",
    "SYN Flag Count",
    "RST Flag Count",
    "PSH Flag Count",
    "ACK Flag Count",
    "URG Flag Count",
    "Down/Up Ratio",
    "Active Mean",
    "Idle Mean"
]

EXPECTED_CLASSES = [
    "BENIGN",
    "Brute Force",
    "DDoS",
    "DoS",
    "Port Scan"
]

# Maximum number of rows used for computationally expensive
# analyses such as correlation and mutual information.
ANALYSIS_SAMPLE_SIZE = 100_000

# Reproducibility
RANDOM_STATE = 42

# Output directory
OUTPUT_DIR = "traffic_characterization"

os.makedirs(OUTPUT_DIR, exist_ok=True)


# ================================================================
# 2. VERIFY DATAFRAME
# ================================================================

if DATAFRAME_NAME not in globals():
    raise NameError(
        f"DataFrame `{DATAFRAME_NAME}` was not found in the notebook. "
        f"Make sure the cell that creates `data_5class` has been executed."
    )

df = data_5class.copy()

print("=" * 80)
print("MLCN 24-FEATURE TRAFFIC CHARACTERIZATION")
print("=" * 80)

print(f"\nDataFrame: {DATAFRAME_NAME}")
print(f"Rows     : {len(df):,}")
print(f"Columns  : {len(df.columns):,}")


# ================================================================
# 3. VERIFY LABEL
# ================================================================

if LABEL_COLUMN not in df.columns:
    raise KeyError(
        f"Label column `{LABEL_COLUMN}` was not found."
    )

missing_features = [f for f in FEATURES if f not in df.columns]

if missing_features:
    raise KeyError(
        "The following required 24 features are missing:\n"
        + "\n".join(f"  - {x}" for x in missing_features)
    )

extra_feature_columns = [
    c for c in df.columns
    if c not in FEATURES and c != LABEL_COLUMN
]

print("\n✓ Label column verified")
print("✓ All 24 required features verified")

print("\nClass distribution:")
print(df[LABEL_COLUMN].value_counts())


# ================================================================
# 4. NUMERIC VALIDATION
# ================================================================

non_numeric = [
    f for f in FEATURES
    if not pd.api.types.is_numeric_dtype(df[f])
]

if non_numeric:
    raise TypeError(
        "Non-numeric feature columns detected:\n"
        + "\n".join(f"  - {x}" for x in non_numeric)
    )

missing_total = int(df[FEATURES].isna().sum().sum())

infinite_total = int(
    np.isinf(
        df[FEATURES].to_numpy(dtype=np.float64)
    ).sum()
)

print("\nData integrity:")
print(f"  Missing feature values : {missing_total:,}")
print(f"  Infinite feature values: {infinite_total:,}")

if missing_total > 0:
    print("⚠ Missing values detected. They will be excluded from statistics.")

if infinite_total > 0:
    print("⚠ Infinite values detected. They will be treated as NaN.")

# Replace infinities only in analysis copy.
analysis_df = df.copy()

analysis_df[FEATURES] = (
    analysis_df[FEATURES]
    .replace([np.inf, -np.inf], np.nan)
)


# ================================================================
# 5. DATA TYPE / RANGE SUMMARY
# ================================================================

dtype_summary = pd.DataFrame({
    "feature": FEATURES,
    "dtype": [str(analysis_df[f].dtype) for f in FEATURES],
    "missing_count": [
        int(analysis_df[f].isna().sum())
        for f in FEATURES
    ],
    "missing_percent": [
        float(analysis_df[f].isna().mean() * 100)
        for f in FEATURES
    ],
    "zero_count": [
        int((analysis_df[f] == 0).sum())
        for f in FEATURES
    ],
    "zero_percent": [
        float((analysis_df[f] == 0).mean() * 100)
        for f in FEATURES
    ],
    "global_min": [
        float(analysis_df[f].min())
        for f in FEATURES
    ],
    "global_max": [
        float(analysis_df[f].max())
        for f in FEATURES
    ]
})

dtype_summary.to_csv(
    os.path.join(OUTPUT_DIR, "01_feature_integrity.csv"),
    index=False
)


# ================================================================
# 6. PER-CLASS FEATURE DISTRIBUTION
# ================================================================
#
# For every feature and every class:
#
#   count
#   min
#   P1
#   P5
#   P10
#   Q1
#   median
#   Q3
#   P90
#   P95
#   P99
#   max
#   mean
#   std
#   skew
#   zero %
#
# ================================================================

percentiles = [
    0.01,
    0.05,
    0.10,
    0.25,
    0.50,
    0.75,
    0.90,
    0.95,
    0.99
]

distribution_rows = []

for class_name in EXPECTED_CLASSES:

    class_df = analysis_df[
        analysis_df[LABEL_COLUMN] == class_name
    ]

    if len(class_df) == 0:
        print(f"⚠ Warning: class `{class_name}` has zero rows.")
        continue

    for feature in FEATURES:

        s = class_df[feature].dropna()

        if len(s) == 0:
            continue

        q = s.quantile(percentiles)

        distribution_rows.append({
            "class": class_name,
            "feature": feature,
            "count": int(s.shape[0]),
            "min": float(s.min()),
            "P1": float(q.loc[0.01]),
            "P5": float(q.loc[0.05]),
            "P10": float(q.loc[0.10]),
            "Q1": float(q.loc[0.25]),
            "median": float(q.loc[0.50]),
            "Q3": float(q.loc[0.75]),
            "P90": float(q.loc[0.90]),
            "P95": float(q.loc[0.95]),
            "P99": float(q.loc[0.99]),
            "max": float(s.max()),
            "mean": float(s.mean()),
            "std": float(s.std(ddof=1)) if len(s) > 1 else 0.0,
            "skew": float(s.skew()) if len(s) > 2 else 0.0,
            "zero_percent": float((s == 0).mean() * 100)
        })

distribution_df = pd.DataFrame(distribution_rows)

distribution_df.to_csv(
    os.path.join(OUTPUT_DIR, "02_per_class_feature_distributions.csv"),
    index=False
)

print("\n✓ Per-class feature distributions generated.")


# ================================================================
# 7. BENIGN VS EACH ATTACK
# ================================================================
#
# For every attack and feature:
#
#   BENIGN median
#   attack median
#   median difference
#   median ratio
#   BENIGN mean
#   attack mean
#   mean difference
#   Cohen's d
#   ROC-AUC
#   direction
#
# ================================================================

benign_df = analysis_df[
    analysis_df[LABEL_COLUMN] == "BENIGN"
]

comparison_rows = []

for attack in EXPECTED_CLASSES:

    if attack == "BENIGN":
        continue

    attack_df = analysis_df[
        analysis_df[LABEL_COLUMN] == attack
    ]

    if len(attack_df) == 0:
        continue

    for feature in FEATURES:

        benign_values = benign_df[feature].dropna()
        attack_values = attack_df[feature].dropna()

        if len(benign_values) == 0 or len(attack_values) == 0:
            continue

        benign_mean = float(benign_values.mean())
        attack_mean = float(attack_values.mean())

        benign_median = float(benign_values.median())
        attack_median = float(attack_values.median())

        median_difference = attack_median - benign_median

        if benign_median != 0:
            median_ratio = attack_median / benign_median
        else:
            median_ratio = np.nan

        # --------------------------------------------------------
        # Cohen's d
        # --------------------------------------------------------

        n1 = len(benign_values)
        n2 = len(attack_values)

        var1 = benign_values.var(ddof=1)
        var2 = attack_values.var(ddof=1)

        pooled_denominator = (
            (n1 + n2 - 2)
        )

        if (
            pooled_denominator > 0
            and np.isfinite(var1)
            and np.isfinite(var2)
        ):
            pooled_variance = (
                ((n1 - 1) * var1)
                + ((n2 - 1) * var2)
            ) / pooled_denominator

            pooled_std = np.sqrt(max(pooled_variance, 0))

            if pooled_std > 0:
                cohens_d = (
                    attack_mean - benign_mean
                ) / pooled_std
            else:
                cohens_d = 0.0
        else:
            cohens_d = 0.0

        # --------------------------------------------------------
        # ROC-AUC
        #
        # BENIGN = 0
        # ATTACK = 1
        #
        # AUC close to 1 or 0 indicates separation.
        # We also calculate absolute AUC.
        # --------------------------------------------------------

        combined_values = pd.concat(
            [benign_values, attack_values],
            ignore_index=True
        )

        combined_labels = np.concatenate([
            np.zeros(len(benign_values), dtype=int),
            np.ones(len(attack_values), dtype=int)
        ])

        try:
            auc = float(
                roc_auc_score(
                    combined_labels,
                    combined_values
                )
            )
        except Exception:
            auc = np.nan

        if np.isfinite(auc):
            separation_auc = max(auc, 1.0 - auc)
        else:
            separation_auc = np.nan

        if attack_median > benign_median:
            direction = "HIGHER_IN_ATTACK"
        elif attack_median < benign_median:
            direction = "LOWER_IN_ATTACK"
        else:
            direction = "SIMILAR"

        comparison_rows.append({
            "attack": attack,
            "feature": feature,
            "benign_mean": benign_mean,
            "attack_mean": attack_mean,
            "mean_difference": attack_mean - benign_mean,
            "benign_median": benign_median,
            "attack_median": attack_median,
            "median_difference": median_difference,
            "median_ratio": median_ratio,
            "cohens_d": float(cohens_d),
            "roc_auc": auc,
            "separation_auc": separation_auc,
            "direction": direction
        })

comparison_df = pd.DataFrame(comparison_rows)

comparison_df.to_csv(
    os.path.join(OUTPUT_DIR, "03_benign_vs_attack_comparison.csv"),
    index=False
)

print("✓ BENIGN-vs-attack comparisons generated.")


# ================================================================
# 8. TOP DISCRIMINATIVE FEATURES PER ATTACK
# ================================================================

top_features_rows = []

for attack in EXPECTED_CLASSES:

    if attack == "BENIGN":
        continue

    temp = comparison_df[
        comparison_df["attack"] == attack
    ].copy()

    if temp.empty:
        continue

    temp["abs_cohens_d"] = temp["cohens_d"].abs()

    temp = temp.sort_values(
        ["separation_auc", "abs_cohens_d"],
        ascending=False
    )

    for rank, (_, row) in enumerate(
        temp.head(24).iterrows(),
        start=1
    ):

        top_features_rows.append({
            "attack": attack,
            "rank": rank,
            "feature": row["feature"],
            "separation_auc": row["separation_auc"],
            "cohens_d": row["cohens_d"],
            "direction": row["direction"],
            "median_difference": row["median_difference"],
            "median_ratio": row["median_ratio"]
        })

top_features_df = pd.DataFrame(top_features_rows)

top_features_df.to_csv(
    os.path.join(OUTPUT_DIR, "04_top_features_per_attack.csv"),
    index=False
)

print("✓ Attack-specific discriminative feature ranking generated.")


# ================================================================
# 9. GLOBAL FEATURE IMPORTANCE USING MUTUAL INFORMATION
# ================================================================
#
# IMPORTANT:
# This does NOT retrain XGBoost.
#
# Mutual information measures nonlinear dependence between
# each feature and the class label.
#
# A maximum of 100,000 rows is used for computational efficiency.
#
# ================================================================

if len(analysis_df) > ANALYSIS_SAMPLE_SIZE:

    mi_df = analysis_df.sample(
        n=ANALYSIS_SAMPLE_SIZE,
        random_state=RANDOM_STATE
    )

else:

    mi_df = analysis_df.copy()

mi_df = mi_df.dropna(
    subset=FEATURES + [LABEL_COLUMN]
)

X_mi = mi_df[FEATURES].to_numpy(dtype=np.float64)

class_to_id = {
    class_name: idx
    for idx, class_name in enumerate(EXPECTED_CLASSES)
}

y_mi = (
    mi_df[LABEL_COLUMN]
    .map(class_to_id)
    .to_numpy()
)

# Replace any remaining non-finite values.
finite_mask = np.isfinite(X_mi).all(axis=1)

X_mi = X_mi[finite_mask]
y_mi = y_mi[finite_mask]

print(
    f"\nCalculating mutual information using "
    f"{len(X_mi):,} samples..."
)

mi_values = mutual_info_classif(
    X_mi,
    y_mi,
    random_state=RANDOM_STATE
)

mi_importance_df = pd.DataFrame({
    "feature": FEATURES,
    "mutual_information": mi_values
})

mi_importance_df = mi_importance_df.sort_values(
    "mutual_information",
    ascending=False
).reset_index(drop=True)

mi_importance_df.insert(
    0,
    "rank",
    np.arange(1, len(mi_importance_df) + 1)
)

mi_importance_df.to_csv(
    os.path.join(OUTPUT_DIR, "05_mutual_information_importance.csv"),
    index=False
)

print("✓ Mutual-information feature importance generated.")


# ================================================================
# 10. CORRELATION MATRIX
# ================================================================

corr_sample = analysis_df[
    FEATURES
].dropna()

if len(corr_sample) > ANALYSIS_SAMPLE_SIZE:

    corr_sample = corr_sample.sample(
        n=ANALYSIS_SAMPLE_SIZE,
        random_state=RANDOM_STATE
    )

global_corr = corr_sample.corr(
    method="spearman"
)

global_corr.to_csv(
    os.path.join(
        OUTPUT_DIR,
        "06_global_spearman_correlation.csv"
    )
)

print("✓ Global Spearman correlation matrix generated.")


# ================================================================
# 11. PER-CLASS CORRELATION MATRICES
# ================================================================

for class_name in EXPECTED_CLASSES:

    class_sample = analysis_df[
        analysis_df[LABEL_COLUMN] == class_name
    ][FEATURES].dropna()

    if len(class_sample) > ANALYSIS_SAMPLE_SIZE:

        class_sample = class_sample.sample(
            n=ANALYSIS_SAMPLE_SIZE,
            random_state=RANDOM_STATE
        )

    if len(class_sample) >= 3:

        class_corr = class_sample.corr(
            method="spearman"
        )

        safe_name = (
            class_name
            .lower()
            .replace(" ", "_")
        )

        class_corr.to_csv(
            os.path.join(
                OUTPUT_DIR,
                f"07_correlation_{safe_name}.csv"
            )
        )

print("✓ Per-class correlation matrices generated.")


# ================================================================
# 12. STRONG FEATURE RELATIONSHIPS
# ================================================================

upper_triangle = global_corr.where(
    np.triu(
        np.ones(global_corr.shape),
        k=1
    ).astype(bool)
)

correlation_rows = []

for feature_a in upper_triangle.index:

    for feature_b in upper_triangle.columns:

        value = upper_triangle.loc[
            feature_a,
            feature_b
        ]

        if pd.notna(value):

            correlation_rows.append({
                "feature_a": feature_a,
                "feature_b": feature_b,
                "spearman_correlation": float(value),
                "absolute_correlation": float(abs(value))
            })

strong_correlations_df = pd.DataFrame(
    correlation_rows
).sort_values(
    "absolute_correlation",
    ascending=False
)

strong_correlations_df.to_csv(
    os.path.join(
        OUTPUT_DIR,
        "08_feature_relationships.csv"
    ),
    index=False
)

print("✓ Feature relationship ranking generated.")


# ================================================================
# 13. ATTACK FINGERPRINT
# ================================================================
#
# We classify feature behavior relative to BENIGN using:
#
#   HIGHER_IN_ATTACK
#   LOWER_IN_ATTACK
#   SIMILAR
#
# We then retain the most strongly separated features.
# ================================================================

fingerprint_rows = []

for attack in EXPECTED_CLASSES:

    if attack == "BENIGN":
        continue

    temp = comparison_df[
        comparison_df["attack"] == attack
    ].copy()

    temp["abs_cohens_d"] = temp["cohens_d"].abs()

    temp = temp.sort_values(
        ["separation_auc", "abs_cohens_d"],
        ascending=False
    )

    # Keep strongest features.
    selected = temp.head(12)

    for rank, (_, row) in enumerate(
        selected.iterrows(),
        start=1
    ):

        fingerprint_rows.append({
            "attack": attack,
            "rank": rank,
            "feature": row["feature"],
            "direction": row["direction"],
            "benign_median": row["benign_median"],
            "attack_median": row["attack_median"],
            "median_difference": row["median_difference"],
            "median_ratio": row["median_ratio"],
            "cohens_d": row["cohens_d"],
            "separation_auc": row["separation_auc"]
        })

fingerprint_df = pd.DataFrame(
    fingerprint_rows
)

fingerprint_df.to_csv(
    os.path.join(
        OUTPUT_DIR,
        "09_attack_fingerprints.csv"
    ),
    index=False
)

print("✓ Attack fingerprints generated.")


# ================================================================
# 14. FEATURE -> TRAFFIC CHARACTERISTIC MAP
# ================================================================
#
# This is an INTERPRETIVE mapping based on your schema.
#
# It is NOT claiming that one traffic variable uniquely determines
# a feature. Several features are mathematically coupled.
#
# The mapping will later be validated experimentally.
# ================================================================

traffic_mapping = {

    "Flow Duration": {
        "traffic_characteristic":
            "flow/connection lifetime",
        "controllable_dimensions":
            "connection lifetime; session duration"
    },

    "Total Fwd Packets": {
        "traffic_characteristic":
            "number of packets transmitted in forward direction",
        "controllable_dimensions":
            "packets per flow; request/message count"
    },

    "Total Backward Packets": {
        "traffic_characteristic":
            "number of packets returned in reverse direction",
        "controllable_dimensions":
            "response packet count; response behavior"
    },

    "Fwd Packets Length Total": {
        "traffic_characteristic":
            "total forward-direction bytes",
        "controllable_dimensions":
            "forward data volume; payload volume"
    },

    "Bwd Packets Length Total": {
        "traffic_characteristic":
            "total reverse-direction bytes",
        "controllable_dimensions":
            "response data volume; payload volume"
    },

    "Fwd Packet Length Mean": {
        "traffic_characteristic":
            "average forward packet size",
        "controllable_dimensions":
            "forward message/payload size"
    },

    "Bwd Packet Length Mean": {
        "traffic_characteristic":
            "average reverse packet size",
        "controllable_dimensions":
            "response/message size"
    },

    "Flow Bytes/s": {
        "traffic_characteristic":
            "average byte throughput over the flow",
        "controllable_dimensions":
            "data volume; flow duration; transmission rate"
    },

    "Flow Packets/s": {
        "traffic_characteristic":
            "average packet throughput over the flow",
        "controllable_dimensions":
            "packet emission rate; flow duration"
    },

    "Flow IAT Mean": {
        "traffic_characteristic":
            "average time between consecutive flow packets",
        "controllable_dimensions":
            "inter-packet timing"
    },

    "Flow IAT Std": {
        "traffic_characteristic":
            "variability of packet inter-arrival timing",
        "controllable_dimensions":
            "timing variability; burstiness"
    },

    "Fwd IAT Mean": {
        "traffic_characteristic":
            "average forward-direction inter-arrival time",
        "controllable_dimensions":
            "forward packet timing"
    },

    "Bwd IAT Mean": {
        "traffic_characteristic":
            "average reverse-direction inter-arrival time",
        "controllable_dimensions":
            "response timing"
    },

    "Packet Length Mean": {
        "traffic_characteristic":
            "average packet size across the flow",
        "controllable_dimensions":
            "message/payload size; packet-size distribution"
    },

    "Packet Length Std": {
        "traffic_characteristic":
            "variability of packet sizes",
        "controllable_dimensions":
            "payload-size variability; message-size diversity"
    },

    "FIN Flag Count": {
        "traffic_characteristic":
            "TCP connection termination behavior",
        "controllable_dimensions":
            "connection lifecycle"
    },

    "SYN Flag Count": {
        "traffic_characteristic":
            "TCP connection-establishment behavior",
        "controllable_dimensions":
            "connection-attempt behavior"
    },

    "RST Flag Count": {
        "traffic_characteristic":
            "TCP reset behavior",
        "controllable_dimensions":
            "connection failure/reset behavior"
    },

    "PSH Flag Count": {
        "traffic_characteristic":
            "TCP PSH behavior",
        "controllable_dimensions":
            "application-data transmission pattern"
    },

    "ACK Flag Count": {
        "traffic_characteristic":
            "TCP acknowledgement behavior",
        "controllable_dimensions":
            "TCP exchange pattern"
    },

    "URG Flag Count": {
        "traffic_characteristic":
            "TCP urgent-data flag behavior",
        "controllable_dimensions":
            "TCP flag behavior"
    },

    "Down/Up Ratio": {
        "traffic_characteristic":
            "forward/reverse packet asymmetry",
        "controllable_dimensions":
            "request/response balance"
    },

    "Active Mean": {
        "traffic_characteristic":
            "average active-period duration using 5-second activity threshold",
        "controllable_dimensions":
            "burst duration; activity continuity"
    },

    "Idle Mean": {
        "traffic_characteristic":
            "average idle gap exceeding 5 seconds",
        "controllable_dimensions":
            "burst spacing; idle periods"
    }
}

mapping_rows = []

for feature in FEATURES:

    item = traffic_mapping[feature]

    mapping_rows.append({
        "feature": feature,
        "traffic_characteristic":
            item["traffic_characteristic"],
        "controllable_dimensions":
            item["controllable_dimensions"]
    })

mapping_df = pd.DataFrame(mapping_rows)

mapping_df.to_csv(
    os.path.join(
        OUTPUT_DIR,
        "10_feature_to_traffic_mapping.csv"
    ),
    index=False
)


# ================================================================
# 15. ATTACK TRAFFIC PROFILE SUMMARY
# ================================================================
#
# Create a human-readable summary for each attack.
# ================================================================

attack_profiles = {}

for attack in EXPECTED_CLASSES:

    if attack == "BENIGN":
        continue

    temp = fingerprint_df[
        fingerprint_df["attack"] == attack
    ]

    profile = {
        "attack": attack,
        "top_features": []
    }

    for _, row in temp.iterrows():

        profile["top_features"].append({
            "feature": row["feature"],
            "direction_vs_benign": row["direction"],
            "benign_median": (
                None
                if pd.isna(row["benign_median"])
                else float(row["benign_median"])
            ),
            "attack_median": (
                None
                if pd.isna(row["attack_median"])
                else float(row["attack_median"])
            ),
            "median_difference": (
                None
                if pd.isna(row["median_difference"])
                else float(row["median_difference"])
            ),
            "median_ratio": (
                None
                if pd.isna(row["median_ratio"])
                else float(row["median_ratio"])
            ),
            "cohens_d": (
                None
                if pd.isna(row["cohens_d"])
                else float(row["cohens_d"])
            ),
            "separation_auc": (
                None
                if pd.isna(row["separation_auc"])
                else float(row["separation_auc"])
            )
        })

    attack_profiles[attack] = profile


# ================================================================
# 16. MACHINE-READABLE JSON REPORT
# ================================================================

report = {
    "analysis": {
        "dataframe": DATAFRAME_NAME,
        "label_column": LABEL_COLUMN,
        "rows": int(len(df)),
        "feature_count": len(FEATURES),
        "features": FEATURES,
        "classes": EXPECTED_CLASSES,
        "random_state": RANDOM_STATE
    },

    "class_distribution": {
        str(k): int(v)
        for k, v in df[LABEL_COLUMN].value_counts().items()
    },

    "attack_profiles": attack_profiles,

    "feature_to_traffic_mapping": traffic_mapping
}

with open(
    os.path.join(
        OUTPUT_DIR,
        "11_attack_traffic_profiles.json"
    ),
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        report,
        f,
        indent=2,
        ensure_ascii=False
    )


# ================================================================
# 17. CONSOLE SUMMARY
# ================================================================

print("\n")
print("=" * 80)
print("CHARACTERIZATION COMPLETE")
print("=" * 80)

print(f"\nOutput directory:")
print(f"  {os.path.abspath(OUTPUT_DIR)}")

print("\nGenerated files:")

for filename in sorted(os.listdir(OUTPUT_DIR)):
    print(f"  ✓ {filename}")


# ================================================================
# 18. DISPLAY MOST IMPORTANT RESULTS
# ================================================================

print("\n")
print("=" * 80)
print("GLOBAL MUTUAL-INFORMATION FEATURE IMPORTANCE")
print("=" * 80)

print(
    mi_importance_df.to_string(index=False)
)


for attack in EXPECTED_CLASSES:

    if attack == "BENIGN":
        continue

    print("\n")
    print("=" * 80)
    print(f"TOP FEATURES: {attack}")
    print("=" * 80)

    display(
        fingerprint_df[
            fingerprint_df["attack"] == attack
        ][[
            "rank",
            "feature",
            "direction",
            "benign_median",
            "attack_median",
            "cohens_d",
            "separation_auc"
        ]].head(12)
    )


print("\n")
print("=" * 80)
print("STRONGEST FEATURE RELATIONSHIPS")
print("=" * 80)

display(
    strong_correlations_df.head(25)
)


print("\n")
print("=" * 80)
print("NEXT STAGE")
print("=" * 80)

print("""
The statistical characterization is now complete.

Next we should NOT manually invent traffic parameters.

Instead, use the generated attack fingerprints to determine:
  1. dominant feature patterns,
  2. correlated feature groups,
  3. observable traffic characteristics,
  4. experimentally controllable parameters,
  5. safe laboratory traffic profiles.

The resulting traffic must then pass through:
  real traffic -> M1 -> M2 -> M3 -> M4 -> XGBoost

rather than manually constructing the 24-feature vector.
""")

print("=" * 80)
```

```python
# ============================================================
# MODERATE UNDERSAMPLING OF TRAINING DATA
# + CLASS WEIGHTS FOR MULTI-CLASS XGBOOST
# ============================================================

from sklearn.utils.class_weight import compute_class_weight
import numpy as np
import pandas as pd

# ------------------------------------------------------------
# 1. Combine training features and labels temporarily
# ------------------------------------------------------------

train_data = X_train.copy()
train_data['Attack Type'] = y_train.values

# ------------------------------------------------------------
# 2. Keep ALL attack samples
#    Undersample only BENIGN
# ------------------------------------------------------------

benign_target = 150_000

benign_data = train_data[
    train_data['Attack Type'] == 'BENIGN'
]

attack_data = train_data[
    train_data['Attack Type'] != 'BENIGN'
]

# Randomly undersample BENIGN
benign_sampled = benign_data.sample(
    n=benign_target,
    random_state=42
)

# Combine
train_balanced = pd.concat(
    [benign_sampled, attack_data],
    ignore_index=True
)

# Shuffle
train_balanced = train_balanced.sample(
    frac=1,
    random_state=42
).reset_index(drop=True)

# ------------------------------------------------------------
# 3. Separate features and target again
# ------------------------------------------------------------

X_train_balanced = train_balanced[retained_features].copy()
y_train_balanced = train_balanced['Attack Type'].copy()

# ------------------------------------------------------------
# 4. Calculate class weights
# ------------------------------------------------------------

classes = np.unique(y_train_balanced)

weights = compute_class_weight(
    class_weight='balanced',
    classes=classes,
    y=y_train_balanced
)

class_weights = dict(zip(classes, weights))

# ------------------------------------------------------------
# 5. Create sample weights for XGBoost
# ------------------------------------------------------------

sample_weights = y_train_balanced.map(class_weights).values

# ------------------------------------------------------------
# 6. Display results
# ------------------------------------------------------------

print("=" * 70)
print("BALANCED TRAINING DATASET")
print("=" * 70)

print("\nTraining distribution:")
print(y_train_balanced.value_counts())

print("\nTraining percentages:")
print(
    y_train_balanced.value_counts(normalize=True)
    .mul(100)
    .round(3)
)

print("\nClass weights:")
for cls, weight in class_weights.items():
    print(f"{cls:15}: {weight:.4f}")

print(f"\nOriginal training samples : {len(X_train):,}")
print(f"Balanced training samples : {len(X_train_balanced):,}")

print("=" * 70)

# ------------------------------------------------------------
# 7. Verify validation and test remain untouched
# ------------------------------------------------------------

print("\nValidation distribution (UNCHANGED):")
print(y_val.value_counts())

print("\nTest distribution (UNCHANGED):")
print(y_test.value_counts())
```

```python
# ============================================================
# XGBOOST MULTI-CLASS IDS TRAINING — CORRECTED VERSION
# ============================================================

import os
import json
import numpy as np
import xgboost as xgb
from xgboost.callback import EarlyStopping

# ------------------------------------------------------------
# 1. Paths
# ------------------------------------------------------------

MODEL_DIR = r"D:\MLCN Dataset\xgboost"

MODEL_PATH = os.path.join(
    MODEL_DIR,
    "xgboost_ids_multiclass.json"
)

FEATURES_PATH = os.path.join(
    MODEL_DIR,
    "xgboost_ids_features.json"
)

os.makedirs(MODEL_DIR, exist_ok=True)

# ------------------------------------------------------------
# 2. Encode target labels
# ------------------------------------------------------------

class_names = sorted(y_train_balanced.unique())

class_to_id = {
    name: idx
    for idx, name in enumerate(class_names)
}

id_to_class = {
    idx: name
    for name, idx in class_to_id.items()
}

y_train_encoded = y_train_balanced.map(class_to_id).astype(np.int32)
y_val_encoded = y_val.map(class_to_id).astype(np.int32)

# ------------------------------------------------------------
# 3. Verify labels
# ------------------------------------------------------------

if y_train_encoded.isna().any():
    raise ValueError("Unknown class found in training labels.")

if y_val_encoded.isna().any():
    raise ValueError("Unknown class found in validation labels.")

# ------------------------------------------------------------
# 4. Verify feature schema
# ------------------------------------------------------------

if list(X_train_balanced.columns) != retained_features:
    raise ValueError(
        "Training feature order does not match retained_features."
    )

if list(X_val.columns) != retained_features:
    raise ValueError(
        "Validation feature order does not match retained_features."
    )

# ------------------------------------------------------------
# 5. Convert to efficient numeric format
# ------------------------------------------------------------

X_train_xgb = X_train_balanced.astype(np.float32)
X_val_xgb = X_val.astype(np.float32)

y_train_xgb = y_train_encoded.to_numpy()
y_val_xgb = y_val_encoded.to_numpy()

sample_weights_xgb = np.asarray(
    sample_weights,
    dtype=np.float32
)

# ------------------------------------------------------------
# 6. Final safety checks
# ------------------------------------------------------------

if not np.isfinite(X_train_xgb.to_numpy()).all():
    raise ValueError(
        "Training data contains NaN or infinite values."
    )

if not np.isfinite(X_val_xgb.to_numpy()).all():
    raise ValueError(
        "Validation data contains NaN or infinite values."
    )

if len(X_train_xgb) != len(sample_weights_xgb):
    raise ValueError(
        "Sample weights do not match training samples."
    )

num_classes = len(class_names)

# ------------------------------------------------------------
# 7. Display configuration
# ------------------------------------------------------------

print("=" * 70)
print("XGBOOST MULTI-CLASS IDS TRAINING")
print("=" * 70)

print(f"Training samples  : {len(X_train_xgb):,}")
print(f"Validation samples: {len(X_val_xgb):,}")
print(f"Features           : {X_train_xgb.shape[1]}")
print(f"Classes            : {num_classes}")
print(f"Class names        : {class_names}")
print(f"Model path         : {MODEL_PATH}")

print("=" * 70)

# ------------------------------------------------------------
# 8. Create model
# ------------------------------------------------------------

model = xgb.XGBClassifier(
    objective="multi:softprob",
    num_class=num_classes,

    n_estimators=3000,
    learning_rate=0.05,

    max_depth=8,
    min_child_weight=2,

    subsample=0.85,
    colsample_bytree=0.85,

    gamma=0.05,

    reg_alpha=0.05,
    reg_lambda=1.5,

    tree_method="hist",
    device="cpu",

    random_state=42,
    n_jobs=-1,

    eval_metric="mlogloss",

    # IMPORTANT:
    # Explicit early stopping configuration
    callbacks=[
        EarlyStopping(
            rounds=100,
            metric_name="mlogloss",
            data_name="validation_1",
            save_best=True
        )
    ]
)

# ------------------------------------------------------------
# 9. Train
# ------------------------------------------------------------

print("\nStarting training...")
print("Early stopping patience: 100 rounds")
print("-" * 70)

model.fit(
    X_train_xgb,
    y_train_xgb,

    sample_weight=sample_weights_xgb,

    eval_set=[
        (X_train_xgb, y_train_xgb),
        (X_val_xgb, y_val_xgb)
    ],

    verbose=50
)

# ------------------------------------------------------------
# 10. Get training results SAFELY
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("TRAINING COMPLETE")
print("=" * 70)

booster = model.get_booster()

best_iteration = booster.best_iteration
best_score = booster.best_score

print(f"Best iteration : {best_iteration}")
print(f"Best mlogloss  : {best_score}")

# ------------------------------------------------------------
# 11. SAVE MODEL
# ------------------------------------------------------------

model.save_model(MODEL_PATH)

print("\n✓ MODEL SAVED")
print(MODEL_PATH)

# ------------------------------------------------------------
# 12. SAVE FEATURE + LABEL METADATA
# ------------------------------------------------------------

metadata = {
    "features": retained_features,
    "classes": class_names,
    "class_to_id": class_to_id,
    "id_to_class": {
        str(k): v
        for k, v in id_to_class.items()
    },
    "num_classes": num_classes,
    "best_iteration": int(best_iteration),
    "best_mlogloss": float(best_score),
    "xgboost_version": xgb.__version__
}

with open(FEATURES_PATH, "w", encoding="utf-8") as f:
    json.dump(metadata, f, indent=4)

print("\n✓ FEATURE/LABEL METADATA SAVED")
print(FEATURES_PATH)

# ------------------------------------------------------------
# 13. FINAL FILE VERIFICATION
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("FILE VERIFICATION")
print("=" * 70)

if os.path.exists(MODEL_PATH):
    print(f"✓ Model exists:    {MODEL_PATH}")
    print(f"  Size: {os.path.getsize(MODEL_PATH) / (1024**2):.2f} MB")
else:
    raise RuntimeError("MODEL FILE WAS NOT CREATED.")

if os.path.exists(FEATURES_PATH):
    print(f"✓ Metadata exists: {FEATURES_PATH}")
else:
    raise RuntimeError("METADATA FILE WAS NOT CREATED.")

print("=" * 70)
print("XGBOOST TRAINING PIPELINE FINISHED SUCCESSFULLY")
print("=" * 70)
```

```python
# ============================================================
# COMPLETE XGBOOST MULTI-CLASS IDS MODEL EVALUATION
# ============================================================

import os
import json
import numpy as np
import pandas as pd
import xgboost as xgb

from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix,
    log_loss,
    roc_auc_score,
    cohen_kappa_score,
    matthews_corrcoef
)

# ============================================================
# 1. CONFIGURATION
# ============================================================

MODEL_DIR = r"D:\MLCN Dataset\xgboost"

MODEL_PATH = os.path.join(
    MODEL_DIR,
    "xgboost_ids_multiclass.json"
)

METADATA_PATH = os.path.join(
    MODEL_DIR,
    "xgboost_ids_features.json"
)

RESULTS_PATH = os.path.join(
    MODEL_DIR,
    "xgboost_evaluation_results.json"
)

REPORT_PATH = os.path.join(
    MODEL_DIR,
    "xgboost_classification_report.csv"
)

CONFUSION_PATH = os.path.join(
    MODEL_DIR,
    "xgboost_confusion_matrix.csv"
)

PREDICTIONS_PATH = os.path.join(
    MODEL_DIR,
    "xgboost_test_predictions.csv"
)

# ============================================================
# 2. LOAD SAVED MODEL
# ============================================================

print("=" * 80)
print("XGBOOST IDS MODEL EVALUATION")
print("=" * 80)

if not os.path.exists(MODEL_PATH):
    raise FileNotFoundError(
        f"Model not found:\n{MODEL_PATH}"
    )

if not os.path.exists(METADATA_PATH):
    raise FileNotFoundError(
        f"Metadata not found:\n{METADATA_PATH}"
    )

model = xgb.XGBClassifier()
model.load_model(MODEL_PATH)

print("\n✓ Model loaded successfully")
print(f"Model: {MODEL_PATH}")

# ============================================================
# 3. LOAD FEATURE / CLASS METADATA
# ============================================================

with open(METADATA_PATH, "r", encoding="utf-8") as f:
    metadata = json.load(f)

retained_features = metadata["features"]
class_names = metadata["classes"]

class_to_id = metadata["class_to_id"]

print(f"\nFeatures: {len(retained_features)}")
print(f"Classes: {class_names}")

# ============================================================
# 4. VERIFY TEST DATA
# ============================================================

print("\n" + "=" * 80)
print("TEST DATA VALIDATION")
print("=" * 80)

# Verify feature order
if list(X_test.columns) != retained_features:
    raise ValueError(
        "TEST FEATURE ORDER DOES NOT MATCH MODEL FEATURE ORDER."
    )

# Verify number of features
if X_test.shape[1] != len(retained_features):
    raise ValueError(
        "TEST FEATURE COUNT DOES NOT MATCH MODEL."
    )

# Verify classes
unknown_classes = set(y_test.unique()) - set(class_names)

if unknown_classes:
    raise ValueError(
        f"Unknown classes found in test set: {unknown_classes}"
    )

# Check NaN
if X_test.isna().any().any():
    raise ValueError(
        "Test set contains NaN values."
    )

# Check infinity
if not np.isfinite(X_test.to_numpy()).all():
    raise ValueError(
        "Test set contains infinite values."
    )

print(f"Test samples : {len(X_test):,}")
print(f"Test features: {X_test.shape[1]}")
print("✓ Feature schema verified")
print("✓ No NaN values")
print("✓ No infinite values")

# ============================================================
# 5. PREPARE TEST DATA
# ============================================================

X_test_eval = X_test.astype(np.float32)

y_test_encoded = (
    y_test
    .map(class_to_id)
    .astype(np.int32)
    .to_numpy()
)

# ============================================================
# 6. GENERATE PREDICTIONS
# ============================================================

print("\n" + "=" * 80)
print("GENERATING TEST PREDICTIONS")
print("=" * 80)

# Probability for every class
y_pred_proba = model.predict_proba(X_test_eval)

# Predicted class ID
y_pred_encoded = np.argmax(
    y_pred_proba,
    axis=1
)

# Convert IDs back to class names
y_pred = np.array(
    [class_names[i] for i in y_pred_encoded]
)

y_true = y_test.to_numpy()

print("✓ Predictions generated")

# ============================================================
# 7. BASIC METRICS
# ============================================================

accuracy = accuracy_score(
    y_true,
    y_pred
)

balanced_accuracy = balanced_accuracy_score(
    y_true,
    y_pred
)

precision_macro = precision_score(
    y_true,
    y_pred,
    average="macro",
    zero_division=0
)

recall_macro = recall_score(
    y_true,
    y_pred,
    average="macro",
    zero_division=0
)

f1_macro = f1_score(
    y_true,
    y_pred,
    average="macro",
    zero_division=0
)

precision_weighted = precision_score(
    y_true,
    y_pred,
    average="weighted",
    zero_division=0
)

recall_weighted = recall_score(
    y_true,
    y_pred,
    average="weighted",
    zero_division=0
)

f1_weighted = f1_score(
    y_true,
    y_pred,
    average="weighted",
    zero_division=0
)

# ============================================================
# 8. LOG LOSS
# ============================================================

test_logloss = log_loss(
    y_test_encoded,
    y_pred_proba,
    labels=np.arange(len(class_names))
)

# ============================================================
# 9. MULTICLASS ROC-AUC
# ============================================================

try:

    roc_auc_ovr_macro = roc_auc_score(
        y_test_encoded,
        y_pred_proba,
        multi_class="ovr",
        average="macro"
    )

    roc_auc_ovr_weighted = roc_auc_score(
        y_test_encoded,
        y_pred_proba,
        multi_class="ovr",
        average="weighted"
    )

except ValueError:

    roc_auc_ovr_macro = None
    roc_auc_ovr_weighted = None

# ============================================================
# 10. OTHER METRICS
# ============================================================

cohen_kappa = cohen_kappa_score(
    y_true,
    y_pred
)

mcc = matthews_corrcoef(
    y_true,
    y_pred
)

# ============================================================
# 11. PRINT OVERALL RESULTS
# ============================================================

print("\n" + "=" * 80)
print("OVERALL MODEL PERFORMANCE")
print("=" * 80)

print(f"Accuracy                 : {accuracy:.6f} ({accuracy * 100:.2f}%)")
print(f"Balanced Accuracy        : {balanced_accuracy:.6f}")
print(f"Macro Precision          : {precision_macro:.6f}")
print(f"Macro Recall             : {recall_macro:.6f}")
print(f"Macro F1                 : {f1_macro:.6f}")
print(f"Weighted Precision       : {precision_weighted:.6f}")
print(f"Weighted Recall          : {recall_weighted:.6f}")
print(f"Weighted F1              : {f1_weighted:.6f}")
print(f"Test Log Loss            : {test_logloss:.6f}")

if roc_auc_ovr_macro is not None:
    print(f"ROC-AUC OVR Macro        : {roc_auc_ovr_macro:.6f}")
    print(f"ROC-AUC OVR Weighted     : {roc_auc_ovr_weighted:.6f}")

print(f"Cohen's Kappa            : {cohen_kappa:.6f}")
print(f"Matthews Correlation     : {mcc:.6f}")

# ============================================================
# 12. CLASSIFICATION REPORT
# ============================================================

print("\n" + "=" * 80)
print("PER-CLASS PERFORMANCE")
print("=" * 80)

report_dict = classification_report(
    y_true,
    y_pred,
    labels=class_names,
    target_names=class_names,
    output_dict=True,
    zero_division=0
)

report_df = pd.DataFrame(report_dict).transpose()

display(
    report_df.style.format({
        "precision": "{:.4f}",
        "recall": "{:.4f}",
        "f1-score": "{:.4f}",
        "support": "{:.0f}"
    })
)

# ============================================================
# 13. CONFUSION MATRIX
# ============================================================

cm = confusion_matrix(
    y_true,
    y_pred,
    labels=class_names
)

cm_df = pd.DataFrame(
    cm,
    index=class_names,
    columns=class_names
)

print("\n" + "=" * 80)
print("CONFUSION MATRIX")
print("=" * 80)

print(
    "\nRows = Actual | Columns = Predicted\n"
)

display(cm_df)

# ============================================================
# 14. NORMALIZED CONFUSION MATRIX
# ============================================================

cm_normalized = confusion_matrix(
    y_true,
    y_pred,
    labels=class_names,
    normalize="true"
)

cm_normalized_df = pd.DataFrame(
    cm_normalized,
    index=class_names,
    columns=class_names
)

print("\n" + "=" * 80)
print("NORMALIZED CONFUSION MATRIX")
print("=" * 80)

display(
    cm_normalized_df.style.format("{:.4f}")
)

# ============================================================
# 15. PER-CLASS ERROR ANALYSIS
# ============================================================

print("\n" + "=" * 80)
print("PER-CLASS ERROR ANALYSIS")
print("=" * 80)

class_analysis = []

for cls in class_names:

    actual_mask = y_true == cls

    total_actual = actual_mask.sum()

    correct = (
        (y_true == cls) &
        (y_pred == cls)
    ).sum()

    incorrect = total_actual - correct

    recall = (
        correct / total_actual
        if total_actual > 0
        else 0
    )

    class_analysis.append({
        "Class": cls,
        "Actual Samples": int(total_actual),
        "Correct": int(correct),
        "Misclassified": int(incorrect),
        "Recall": recall
    })

class_analysis_df = pd.DataFrame(
    class_analysis
)

display(
    class_analysis_df.style.format({
        "Recall": "{:.4f}"
    })
)

# ============================================================
# 16. PREDICTION CONFIDENCE
# ============================================================

prediction_confidence = np.max(
    y_pred_proba,
    axis=1
)

print("\n" + "=" * 80)
print("PREDICTION CONFIDENCE")
print("=" * 80)

print(
    f"Mean confidence   : {prediction_confidence.mean():.4f}"
)

print(
    f"Median confidence : {np.median(prediction_confidence):.4f}"
)

print(
    f"Minimum confidence: {prediction_confidence.min():.4f}"
)

print(
    f"Maximum confidence: {prediction_confidence.max():.4f}"
)

# ============================================================
# 17. CONFIDENCE ON CORRECT VS INCORRECT PREDICTIONS
# ============================================================

correct_mask = y_true == y_pred

if correct_mask.any():
    correct_confidence = prediction_confidence[correct_mask]

    print(
        f"Correct prediction mean confidence: "
        f"{correct_confidence.mean():.4f}"
    )

if (~correct_mask).any():
    incorrect_confidence = prediction_confidence[~correct_mask]

    print(
        f"Incorrect prediction mean confidence: "
        f"{incorrect_confidence.mean():.4f}"
    )

# ============================================================
# 18. MISCLASSIFIED SAMPLE ANALYSIS
# ============================================================

misclassified_mask = y_true != y_pred

misclassified_count = misclassified_mask.sum()

print("\n" + "=" * 80)
print("MISCLASSIFICATION SUMMARY")
print("=" * 80)

print(
    f"Correct predictions     : {correct_mask.sum():,}"
)

print(
    f"Incorrect predictions   : {misclassified_count:,}"
)

print(
    f"Error rate              : "
    f"{misclassified_count / len(y_true) * 100:.4f}%"
)

# ============================================================
# 19. CREATE PREDICTION DATAFRAME
# ============================================================

prediction_df = X_test.reset_index(drop=True).copy()

prediction_df["Actual"] = y_true
prediction_df["Predicted"] = y_pred
prediction_df["Confidence"] = prediction_confidence

prediction_df["Correct"] = (
    prediction_df["Actual"] ==
    prediction_df["Predicted"]
)

# ============================================================
# 20. FEATURE IMPORTANCE
# ============================================================

print("\n" + "=" * 80)
print("FEATURE IMPORTANCE")
print("=" * 80)

importance_values = model.feature_importances_

feature_importance_df = pd.DataFrame({
    "Feature": retained_features,
    "Importance": importance_values
})

feature_importance_df = (
    feature_importance_df
    .sort_values(
        "Importance",
        ascending=False
    )
    .reset_index(drop=True)
)

display(
    feature_importance_df.style.format({
        "Importance": "{:.6f}"
    })
)

# ============================================================
# 21. SAVE CLASSIFICATION REPORT
# ============================================================

report_df.to_csv(
    REPORT_PATH
)

# ============================================================
# 22. SAVE CONFUSION MATRIX
# ============================================================

cm_df.to_csv(
    CONFUSION_PATH
)

# ============================================================
# 23. SAVE TEST PREDICTIONS
# ============================================================

prediction_df.to_csv(
    PREDICTIONS_PATH,
    index=False
)

# ============================================================
# 24. SAVE EVALUATION SUMMARY
# ============================================================

evaluation_results = {
    "model": "XGBoost Multi-Class IDS",

    "test_samples": int(len(X_test)),

    "num_features": int(len(retained_features)),

    "num_classes": int(len(class_names)),

    "classes": class_names,

    "accuracy": float(accuracy),

    "balanced_accuracy": float(
        balanced_accuracy
    ),

    "macro_precision": float(
        precision_macro
    ),

    "macro_recall": float(
        recall_macro
    ),

    "macro_f1": float(
        f1_macro
    ),

    "weighted_precision": float(
        precision_weighted
    ),

    "weighted_recall": float(
        recall_weighted
    ),

    "weighted_f1": float(
        f1_weighted
    ),

    "log_loss": float(
        test_logloss
    ),

    "roc_auc_ovr_macro": (
        float(roc_auc_ovr_macro)
        if roc_auc_ovr_macro is not None
        else None
    ),

    "roc_auc_ovr_weighted": (
        float(roc_auc_ovr_weighted)
        if roc_auc_ovr_weighted is not None
        else None
    ),

    "cohen_kappa": float(
        cohen_kappa
    ),

    "matthews_correlation_coefficient": float(
        mcc
    ),

    "mean_prediction_confidence": float(
        prediction_confidence.mean()
    ),

    "median_prediction_confidence": float(
        np.median(prediction_confidence)
    ),

    "best_iteration": metadata.get(
        "best_iteration"
    ),

    "best_validation_mlogloss": metadata.get(
        "best_mlogloss"
    ),

    "classification_report": report_dict
}

with open(
    RESULTS_PATH,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        evaluation_results,
        f,
        indent=4
    )

# ============================================================
# 25. FINAL SUMMARY
# ============================================================

print("\n" + "=" * 80)
print("EVALUATION COMPLETE")
print("=" * 80)

print(
    f"\nAccuracy          : {accuracy * 100:.2f}%"
)

print(
    f"Balanced Accuracy : {balanced_accuracy * 100:.2f}%"
)

print(
    f"Macro F1          : {f1_macro:.4f}"
)

print(
    f"Weighted F1       : {f1_weighted:.4f}"
)

print(
    f"Test Log Loss     : {test_logloss:.6f}"
)

if roc_auc_ovr_macro is not None:
    print(
        f"ROC-AUC (Macro)   : {roc_auc_ovr_macro:.4f}"
    )

print(
    f"\nCorrect predictions   : {correct_mask.sum():,}"
)

print(
    f"Incorrect predictions : {misclassified_count:,}"
)

print("\nSaved files:")

print(f"✓ Evaluation summary : {RESULTS_PATH}")
print(f"✓ Classification     : {REPORT_PATH}")
print(f"✓ Confusion matrix   : {CONFUSION_PATH}")
print(f"✓ Predictions        : {PREDICTIONS_PATH}")

print("\n" + "=" * 80)
print("MODEL EVALUATION FINISHED SUCCESSFULLY")
print("=" * 80)
```
